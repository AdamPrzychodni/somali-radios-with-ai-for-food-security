# 03 · Transcription

Transcribe broadcast audio with ElevenLabs Scribe or Gemini.

Needs `pip install -e '.[apis]'` and `ELEVENLABS_API_KEY` / `GEMINI_API_KEY` in a `.env` file (see `.env.example`).

```python
from dotenv import load_dotenv

from somali_foodsec_radio.config import get_config
from somali_foodsec_radio.transcription.asr_elevenlabs import transcribe_with_elevenlabs
from somali_foodsec_radio.transcription.asr_gemini import transcribe_audio
from somali_foodsec_radio.transcription.batch import run_directory_batch

load_dotenv()
cfg = get_config()
```

## ElevenLabs Scribe — single file

```python
audio_file = cfg["paths"]["data_raw"] / "IDAACADDA 01-AUG-2022.mp3"
text = transcribe_with_elevenlabs(audio_file, model_id=cfg["asr"]["elevenlabs_model"])
print(text[:1000])
```

## Gemini — batch-transcribe a directory of MP3s

Writes one `.txt` transcript per broadcast into `data/interim/transcripts/`.

```python
results = run_directory_batch(
    input_dir=str(cfg["paths"]["data_raw"]),
    output_dir=str(cfg["paths"]["transcripts"]),
    work_fn=lambda mp3: transcribe_audio(mp3, model=cfg["asr"]["gemini_model"]),
    input_glob="*.mp3",
    failure_log="failed_transcriptions.json",
)
print(results)
```
