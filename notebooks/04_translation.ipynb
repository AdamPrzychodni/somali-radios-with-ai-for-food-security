# 04 · Translation

Translate Somali transcripts to English with NLLB and MADLAD.

Needs `uv sync --extra analysis`.

```python
from somali_foodsec_radio.config import get_config
from somali_foodsec_radio.translation.pipeline import (
    load_transcription_data,
    run_translation_pipeline,
)

cfg = get_config()
csv_path = cfg["paths"]["transcripts"] / cfg["outputs"]["transcriptions_csv"]
transcriptions_df = load_transcription_data(str(csv_path))
```

## NLLB-200-3.3B

```python
nllb_result = run_translation_pipeline(
    df=transcriptions_df,
    model_name=cfg["translation"]["models"]["nllb_3_3b"],
    model_type="nllb",
    num_rows_to_translate=1,
)
```

## MADLAD-400-3B

```python
madlad_3b_result = run_translation_pipeline(
    df=transcriptions_df,
    model_name=cfg["translation"]["models"]["madlad_3b"],
    model_type="madlad",
    num_rows_to_translate=1,
)
```
