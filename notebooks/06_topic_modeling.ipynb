# 06 · Topic Modelling

Extract themes (BERTopic) and locations (spaCy) from the English transcripts, then pair each theme with an IPC geography.

Needs `uv sync --extra analysis` and the spaCy model:
`python -m spacy download en_core_web_sm`.

```python
from somali_foodsec_radio.config import get_config
from somali_foodsec_radio.topics import run_topic_pipeline

cfg = get_config()
```

```python
theme_map = {int(k): v for k, v in cfg["topics"]["theme_map"].items()}
bertopic_cfg = cfg["topics"]["bertopic"]
model_kwargs = {
    "min_topic_size": bertopic_cfg["min_topic_size"],
    "n_gram_range": tuple(bertopic_cfg["n_gram_range"]),
}

df_pairs = run_topic_pipeline(
    transcripts_dir=str(cfg["paths"]["translations"]),
    geojson_path=str(
        cfg["paths"]["data_external"] / "Somalia-Somalia IPC Post GU 2024.json"
    ),
    theme_map=theme_map,
    model_kwargs=model_kwargs,
    prob_threshold=cfg["topics"]["prob_threshold"],
)
df_pairs
```
